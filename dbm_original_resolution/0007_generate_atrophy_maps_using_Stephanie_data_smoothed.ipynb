```python
%%bash

function calc() { awk "BEGIN{ print $* }"; }
fwhm="4vox"
sigma_num=$(calc "$(echo ${fwhm} | grep -o -E '^[0-9]+')/(2*sqrt(2*log(2)))")
if [[ ${fwhm} =~ [0-9]+mm$ ]]; then
  fwhm_type=1
elif [[ ${fwhm} =~ [0-9]+vox$ ]]; then
  fwhm_type=0
else
  failure "Parse error for FWHM entry \"${fwhm}\", must end with vox or mm"
fi

echo "FWHM: ${fwhm}, sigma: ${sigma_num}, type: ${fwhm_type}"

# smooth the images in the Stephanie_data_smoothed folder using fslmaths with a 4vox kernel
stephanie_dir="/Users/aeed/Documents/Work/M83_clearing/Manuscript_analysis/dbm_output_original_resolution/Stephanie_data_smoothed"
for img in ${stephanie_dir}/*.nii.gz; do
    echo "Smoothing $img"
    img_stem=$(remove_ext $img)
    img_stem=$(basename $img_stem)
    echo "Output will be ${img_stem}_smoothed.nii.gz"

    SmoothImage 3 $img  ${sigma_num} ${stephanie_dir}/${img_stem}_fwhm_${fwhm}.nii.gz ${fwhm_type} 0
done
```

## get average values of the atrophy maps in each ROI


```python
import numpy as np
import pandas as pd
import nibabel as nib
import os

DEFAULT_ATLAS     = '/Users/aeed/Documents/Work/M83_clearing/Manuscript_analysis/dbm_output_original_resolution/final-target/ABA_51_bilateral_resampled.nii.gz'
DEFAULT_ATLAS_CSV = '/Users/aeed/Documents/Work/RABIES_templates/ROIs_51_bilateral_interleaved_RABIES_labels.csv'

def extract_roi_means(
    img_path,
    output_dir,
    atlas_path=DEFAULT_ATLAS,
    atlas_csv=DEFAULT_ATLAS_CSV,
):
    # Load images
    img_data   = nib.load(img_path).get_fdata()
    atlas_data = nib.load(atlas_path).get_fdata()
    labels_df  = pd.read_csv(atlas_csv)

    # Compute average voxel value per ROI
    means = []
    for _, row in labels_df.iterrows():
        mask     = atlas_data == row['label_value']
        mean_val = img_data[mask].mean() if mask.any() else np.nan
        means.append(mean_val)

    # Build output dataframe
    out_df = pd.DataFrame({
        'index':            labels_df['label_value'].values,
        'name':             labels_df['name'].values,
        'mean_voxel_value': means,
    })

    # Save
    os.makedirs(output_dir, exist_ok=True)
    img_stem  = os.path.basename(img_path).replace('.nii.gz', '').replace('.nii', '')
    out_path  = os.path.join(output_dir, f'{img_stem}_roi_means.tsv')
    out_df.to_csv(out_path, sep='\t', index=False)

    return out_path
```

```python
# loop over all the pff atrophy maps and extract the mean values in each ROI
stephanie_atrophy_dir = "/Users/aeed/Documents/Work/M83_clearing/Manuscript_analysis/dbm_output_original_resolution/Stephanie_data_smoothed/atrophy_maps"
PFF=['sub-hM8362334086041083M6',
    'sub-hM836323993941035M4',
    'sub-hM8363224086001085M5',
    'sub-hM836373993921113M9',
    'sub-hM8362324086031084M5']
for pff in PFF:
    img_path = f"{stephanie_atrophy_dir}/{pff}_atrophy_map_masked.nii.gz"
    extract_roi_means(img_path, stephanie_atrophy_dir)
```
