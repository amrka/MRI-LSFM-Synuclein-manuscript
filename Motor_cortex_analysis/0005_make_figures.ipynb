```python
import glob
import shutil
from pathlib import Path
```

```python
orig_dir = "/Users/aeed/Documents/Work/M83_clearing/Manuscript_analysis/Motor_cortex_dgamma_5_smooth_just_motion"
zstats_perc_lvl_1_dir = f"{orig_dir}/zstats_perc_lvl_1_dir"
```

```python
subjects_list = ['hM835603942971011M8',
'hM8362824086011084M8',
'hM836223979691084M1',
'hM8362224086051084M2',
'hM8362234086021084M3',
'hM8362334086041083M6',
'hM836323993941035M4',
'hM8363224086001085M5',
'hM836373993921113M9',
'hM8362324086031084M5',]
```

```python
if not os.path.exists(zstats_perc_lvl_1_dir):
    os.makedirs(zstats_perc_lvl_1_dir, exist_ok=True)

for i in np.arange(1,5):
    os.makedirs((f"{zstats_perc_lvl_1_dir}/zstat{i}"), exist_ok=True)
```

```python
# if you used, it will work only wiht ! and %%bash, not inside python cells
import os
os.environ["zstats_perc_lvl_1_dir"] = f"{orig_dir}/zstats_perc_lvl_1_dir"
```

```python
%%bash
for i in {1..4}; do
    mkdir -p "$zstats_perc_lvl_1_dir/zstat$i"

    for img in /Users/aeed/Documents/Work/M83_clearing/Manuscript_analysis/Motor_cortex_dgamma_5_smooth_just_motion/Motor_cortex_activation_1st_level_WorkingDir/Motor_cortex_activation_1st_level/_run_id_?_session_id_01_subject_id_*/Fit_Design_to_Timeseries/results/zstat${i}.nii.gz;do

        folder=$(basename "$(dirname "$(dirname "$(dirname "$img")")")")

        run=${folder#_run_id_}
        run=${run%%_*}

        subject=${folder##*_subject_id_}

        echo "Subject: $subject   Run: $run"
        fslmaths "$img" -thr 1.96 -bin  "$zstats_perc_lvl_1_dir/zstat${i}/sub-${subject}_run-${run}_zstat${i}.nii.gz"
    done
done

```

```python
# for each dir sum the images and divide by the number of images to get the percentage of subjects that have activation in each voxel
for i in np.arange(1,5):
    zstat_dir = f"{zstats_perc_lvl_1_dir}/zstat{i}"
    output_img = f"{zstats_perc_lvl_1_dir}/zstat{i}_percentage.nii.gz"
    img_list = glob.glob(f"{zstat_dir}/*.nii.gz")
    num_imgs = len(img_list)
    if num_imgs == 0:
        print(f"No images found in {zstat_dir}")
        continue
    sum_img = None
    for img in img_list:
        img_data = nib.load(img).get_fdata()
        if sum_img is None:
            sum_img = np.zeros_like(img_data)
        sum_img += img_data
    percentage_img = sum_img / num_imgs * 100
    # get the header and affine from one of the images    sample_img = nib.load(img_list[0])
    img_obj = nib.load(img)
    header = img_obj.header
    affine = img_obj.affine
    nib.save(nib.Nifti1Image(percentage_img, affine, header), output_img)
```
